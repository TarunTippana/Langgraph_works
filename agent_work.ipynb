```python
import os 
from dotenv import load_dotenv
import warnings
warnings.filterwarnings("ignore")
```

```python
load_dotenv()
apikey=os.getenv("MISTRAL_API_KEY")
if apikey:
    print("Mistral API key Loaded successfully")
else: 
    print("Mistral API Not loaded")    
```

```python
from langchain_mistralai import ChatMistralAI
"""Mistral Small 4
Our powerful hybrid model unifying instruct, reasoning, and coding capabilities in a single model. 119B parameters with 6.5B active. context 256k """
llm = ChatMistralAI(model="mistral-small-2603")
```

```python
from langchain_core.tools import tool
import requests
```

```python
from langchain_community.tools import DuckDuckGoSearchRun
search_tool = DuckDuckGoSearchRun()
results = search_tool.invoke("tell me a joke on girls",timeout=20)
```

```python
print(results)
```

```python
from langsmith import Client
from langchain_classic.agents import create_react_agent, AgentExecutor
client = Client()
prompt = client.pull_prompt("hwchase17/react",dangerously_pull_public_prompt=True)
```

```python
print(prompt.template)
```

```python
agent = create_react_agent(llm=llm,
                           tools = [search_tool],
                           prompt = prompt)
```

```python
agent_executor = AgentExecutor(agent = agent,
                               tools = [search_tool],
                               verbose = True ,
                               handle_parsing_errors=True,
                               stop=)
```

```python
response = agent_executor.invoke({"input":"give a five pointer news for today"})
print(response)
```
